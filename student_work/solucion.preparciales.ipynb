```python
def minmax(L):
    if len(L)==1:
        min=L[0]
        max=L[0]
        return min,max
    else:
        if minmax(L[:-1])<L[-1]:
            return L[-1]
        else:
            return minmax(L[:-1])

        if minmax(L[:-1])>L[-1]:
            return L[-1]
        else:
            return minmax(L[:-1])
```

```python
def picos(L):
    if len(L)<3:
        return []
    if L[0]<L[1]>L[2]:
        return [L[1]]+picos(L[1:])
    else:
        return picos(L[1:])
            
```
